```python
from arc import *
from numpy import pi,sqrt
# use ARC
print(getCitationForARC())
atom=Rubidium87()
```

**First step. We calculate matrix element using hyperfine structure**

```python
P=0.8e-6 #Power
w0=50e-6 #beam waist
cl= 299792458 #light velocity
fpieps0=1.11265005620e-10 #4 pi eps0
Intensity=2*P/(pi*w0**2) #Peak intensity of gaussian beam
El= sqrt(8*pi*Intensity/(cl*fpieps0)) #Electric field in V/m
hp=1.054571817e-34 #Plancks constant
ec= 1.602176634e-19 #elementary charge
aB= 5.29177210544e-11 #Bohr radius 
dip=atom.getDipoleMatrixElementHFS(5,0, 0.5, 2, 0, 5, 1, 1.5, 3, 0, 0, 0.5)
dip*El*ec*aB/(2*pi*hp)*1e-6
```

**Second step. We calculate matrix element using hyperfine structure**

```python
P=0.7e-3 #Power
w0=5e-6 #beam waist
cl= 299792458 #light velocity
fpieps0=1.11265005620e-10 #4 pi eps0
Intensity=2*P/(pi*w0**2)
El= sqrt(8*pi*Intensity/(cl*fpieps0))
hp=1.054571817e-34 #Plancks constant
ec= 1.602176634e-19 #elementary charge
aB= 5.29177210544e-11 #Bohr radius 
dip=atom.getDipoleMatrixElementHFS(5,1, 1.5, 3, 0, 7, 0, 0.5, 2, 0, 0, 0.5)
dip*El*ec*aB/(2*pi*hp)*1e-6
```

**Third step. Here we use only fine structure. Can be improved**

```python
P=20e-3 #Power
w0=5e-6 #beam waist
cl= 299792458 #light velocity
fpieps0=1.11265005620e-10 #4 pi eps0
Intensity=2*P/(pi*w0**2)
El= sqrt(8*pi*Intensity/(cl*fpieps0))
hp=1.054571817e-34 #Plancks constant
ec= 1.602176634e-19 #elementary charge
aB= 5.29177210544e-11 #Bohr radius 
dip=atom.getDipoleMatrixElement(6, 0, 0.5, 0.5, 37,1, 1.5, 0.5, 0)
dip*El*ec*aB/(2*pi*hp)*1e-6
```
